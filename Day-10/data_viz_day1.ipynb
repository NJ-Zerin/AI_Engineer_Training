```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create dataset
data = {
    'Student': [
        'Alice', 'Bob', 'Charlie', 'David', 'Eva',
        'Frank', 'Grace', 'Hannah', 'Ian', 'Julia',
        'Kevin', 'Laura', 'Mike', 'Nina', 'Oscar',
        'Paula', 'Quinn', 'Rachel', 'Steve', 'Tina'
    ],
    'Math': [
        85, 90, 78, 92, 88,
        74, 81, 89, 76, 94,
        87, 82, 79, 91, 84,
        86, 88, 90, 77, 93
    ],
    'Science': [
        80, 95, 75, 89, 92,
        78, 84, 91, 73, 96,
        85, 83, 77, 90, 88,
        87, 89, 92, 76, 94
    ],
    'English': [
        78, 85, 82, 88, 90,
        75, 80, 87, 79, 91,
        84, 81, 76, 89, 85,
        83, 86, 88, 78, 92
    ]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# ---------- Line Chart (Math)----------
plt.figure(figsize=(16,8), dpi = 120)
plt.plot(df['Student'], df['Math'], marker='o', label='Math', color = 'blue')
plt.title('Students Marks (Math)', fontsize=16)

plt.xlabel('Students')
plt.ylabel('Marks')

plt.xticks(rotation=45)
plt.legend(loc='upper left', fontsize=12) 
plt.grid(True)
plt.show()
```

```python
# ---------- Line Chart (Science)----------
plt.figure(figsize=(16,8), dpi = 120)
plt.plot(df['Student'], df['Science'], marker='s', label='Science', color = 'green')
plt.title('Students Marks (Science)', fontsize=16)

plt.xlabel('Students')
plt.ylabel('Marks')

plt.xticks(rotation=45)
plt.legend(loc='upper left', fontsize=12)  
plt.grid(True)

plt.show()

```

```python
# ---------- Line Chart (English)----------
plt.figure(figsize=(16,8), dpi = 120)
plt.plot(df['Student'], df['English'], marker='^', label='English', color = 'purple')
plt.title('Students Marks (English)', fontsize=16)

plt.xlabel('Students')
plt.ylabel('Marks')

plt.xticks(rotation=45)
plt.legend(loc='upper left', fontsize=12)  
plt.grid(True)

plt.show()
```

```python
# ---------- Line Chart (All subjects)----------
plt.figure(figsize=(16,8), dpi = 120)

plt.plot(df['Student'], df['Math'], marker='o', label='Math')
plt.plot(df['Student'], df['Science'], marker='s', label='Science')
plt.plot(df['Student'], df['English'], marker='^', label='English')
plt.title('Students Marks Comparison Line Chart', fontsize=16)

plt.xlabel('Students')
plt.ylabel('Marks')

plt.legend(loc='upper left', fontsize=12)
plt.grid(True)

plt.savefig("line.png")

plt.show()
```

```python
# ---------- Bar Chart (Math)----------
plt.figure(figsize=(16,8), dpi = 120)
x = range(len(df['Student']))

plt.bar(x, df['Math'], label='Math', color = 'blue')

plt.title('Math Marks Bar Chart', fontsize=16)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.xticks(x, df['Student'], rotation=45)
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()

```

```python
# ---------- Bar Chart (Science)----------
plt.figure(figsize=(16,8), dpi = 120)
x = range(len(df['Science']))

plt.bar(x, df['Science'], label='Science', color = 'green')

plt.title('Science Marks Bar Chart', fontsize=16)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.xticks(x, df['Student'], rotation=45)
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ---------- Bar Chart (English)----------
plt.figure(figsize=(16,8), dpi = 120)
x = range(len(df['English']))

plt.bar(x, df['English'], label='English', color = 'purple')

plt.title('English Marks Bar Chart', fontsize=16)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.xticks(x, df['Student'], rotation=45)
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ---------- Bar Chart (All subjects)----------
plt.figure(figsize=(16,8), dpi = 120)

bar_width = 0.25
x = range(len(df['Student']))

plt.bar(x, df['Math'], width = bar_width, label='Math')
plt.bar([i + bar_width for i in x], df['Science'], width = bar_width, label='Science')
plt.bar([i + bar_width*2 for i in x], df['English'], width = bar_width, label='English')

plt.title('Students Marks Comparison Bar Chart', fontsize=16)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.xticks([i + bar_width for i in x], df['Student'], rotation=45)
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.savefig("Bar_plot.png")

plt.show()
```

```python
# ---------- Histogram (Math marks)----------
plt.figure(figsize=(16,8), dpi=120)

plt.hist(df['Math'], bins=5, label='Math', color = 'blue')

plt.title('Math Marks Distribution Histogram', fontsize=16)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ---------- Histogram (Science marks)----------
plt.figure(figsize=(16,8), dpi=120)

plt.hist(df['Science'], bins=5, label='Science', color = 'green')

plt.title('Science Marks Distribution Histogram', fontsize=16)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ---------- Histogram (English marks)----------
plt.figure(figsize=(16,8), dpi=120)

plt.hist(df['English'], bins=5,label='English', color = 'purple')

plt.title('English Marks Distribution Histogram', fontsize=16)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ---------- Pie Chart ---------- #
# Average marks per subject #
avg_marks = df[['Math', 'Science', 'English']].mean() 

plt.figure(figsize=(16,8), dpi=120) 
plt.pie(avg_marks, labels=avg_marks.index, autopct='%1.1f%%', startangle=90) 

plt.title('Average Marks Pie Chart') 
plt.savefig("pie_chart.png")

plt.show()
```
