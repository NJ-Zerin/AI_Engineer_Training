```python
import numpy as np
from scipy import stats
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
sample_mean = np.mean(sample)
sample_std = np.std(sample,ddof = 1)
n = len(sample)
population_mean = 170
```

```python
t_stast = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
t_stast = float(t_stast)
t_stast
```

```python
p_value = 2 * ( 1 - stats.t.cdf(abs(t_stast), df = 1))
p_value = float (p_value)
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("I will reject the null hypothesis")
else:
    print("I will accept the null hypothesis")
    
```
