```python
# Filter students with math marks ≥ 80
math_marks = df[df['math'] >= 80]
math_marks
```

```python
# Calculate average marks per subject
average_marks = df[['math', 'english', 'science']].agg('mean')
average_marks
```

```python
# Sort students by science marks (descending)
sorted_science = df[['name', 'science']].sort_values(by='science', ascending=False)
sorted_science
```

```python
df.describe()
```

```python
# Filter students who scored less than 60 in any subject
less_than_60 = df[(df['math'] < 60) | (df['english'] < 60) | (df['science'] < 60)]
less_than_60
```

```python
# Count how many students scored above 80 in English
high_english_marks = df[df['english'] > 80][['id', 'name', 'english']]
high_english_marks
```
