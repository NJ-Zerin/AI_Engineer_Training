```python
#Import
import numpy as np 
import pandas as pd
```

```python
#Load data
df = pd.read_csv('students_data.csv')
```

```python
#Show the datas 
df
```

```python
#Show the information about the data
df.info()
```

```python
#Show the shape of the data
df.shape
```

```python
#Calculate total marks per student 
df['Total_Marks'] = df[['math', 'English', 'science']].sum(axis=1)
```

```python
#Display the updated DataFrame
df
```

```python
#Calculate Avg marks per student
df['Average_Marks'] = df[['math', 'English', 'science']].mean(axis=1).round(2)
```

```python
#Display updated DataFrame
df
```

```python
#Students with attendance ≥ 90%
df.loc[df['attendance'] >= 90, ['student_id', 'name', 'attendance']]
```

```python
# Students with attendance < 80%
df.loc[df['attendance'] < 80, ['student_id', 'name', 'attendance']]
```

```python
#Students high attendance (≥ 65%)
high_attendance = df.loc[df["attendance"] >= 65, ['student_id', 'name', 'attendance']]
high_attendance
```

```python
#Students low attendance (< 65)
low_attendance  = df.loc[df["attendance"] < 65, ['student_id', 'name', 'attendance']]
low_attendance
```

```python
#Students each subject totals
subject_totals = df[['math', 'English', 'science']].sum()
```

```python
# Create new row "Total Each Subject" with student_id and student_name as empty
total_each_marks = pd.DataFrame([{
    'student_id': '',
    'name': 'Total Each Subject',
    'math': subject_totals['math'],
    'English': subject_totals['English'],
    'science': subject_totals['science']
}])
```

```python
#Students each subject averages
each_subject_averages = df[['math', 'English', 'science']].mean().round(2)
```

```python
# Create new row "Average Each Subject"
each_avg_marks = pd.DataFrame([{
    'student_id': '',
    'name': 'Average Each Subject',
    'math': each_subject_averages['math'],
    'English': each_subject_averages['English'],
    'science': each_subject_averages['science']
}])
```

```python
#Combine original DataFrame with total and average rows
df_each_subject = pd.concat([df, total_each_marks, each_avg_marks], ignore_index=True)
```

```python
#Display the updated DataFrame
df_each_subject
```

```python
#Overall class average
overall_class_avg = float(df['Average_Marks'].mean().round(2))
```

```python
overall_class_avg
```

```python
# Only consider actual students
df_students_only = df_each_subject.iloc[:-2].copy()
```

```python
# Top-performing student
top_student = df_students_only.loc[
    df_students_only['Total_Marks'].idxmax(),
    ['student_id', 'name', 'Total_Marks', 'Average_Marks']
]
```

```python
top_student
```

```python
# Lowest-performing student
lowest_student = df_students_only.loc[
    df_students_only['Total_Marks'].idxmin(),
    ['student_id', 'name', 'Total_Marks', 'Average_Marks']
]
```

```python
lowest_student
```

```python
# Sort students by Average_Marks (descending)
df_sorted = df.sort_values(by='Average_Marks', ascending=False).reset_index(drop=True)
```

```python
df_sorted
```

```python
# Generate Rank column
df_sorted['Rank'] = df_sorted['Average_Marks'].rank(method='min', ascending=False).astype(int)
```

```python
df_sorted
```

```python
# Generate Rank column
df_sorted['Rank'] = df_sorted['Average_Marks'].rank(method='min', ascending=False).astype(int)
```

```python
# 7️⃣ Prepare extra rows with NaN for Total_Marks, Average_Marks, Rank
extra_rows = df_each_subject.iloc[-2:].copy()
extra_rows['Total_Marks'] = pd.NA
extra_rows['Average_Marks'] = pd.NA
extra_rows['Rank'] = pd.NA
```

```python
#Combine sorted students with extra rows
final_report = pd.concat([df_sorted, total_each_marks, each_avg_marks], ignore_index=True)
```

```python
#Convert Rank column to Int64 (allows NaN in extra rows but keeps integers for students)
final_report['Rank'] = final_report['Rank'].astype('Int64')
```

```python
#Export to CSV
final_report.to_csv('student_report.csv', index=False)
```
